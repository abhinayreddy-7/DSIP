```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read Image 1
src_image = cv2.imread('/content/Image1_LowContrast.jpg',
                       cv2.IMREAD_GRAYSCALE)

# Check whether image is loaded
if src_image is None:
    print("Image not found!")
else:

    # -------------------------
    # Contrast Stretching
    # -------------------------
    min_val = np.min(src_image)
    max_val = np.max(src_image)

    enhanced_image = ((src_image - min_val) /
                      (max_val - min_val) * 255).astype(np.uint8)

    # -------------------------
    # Display Original & Enhanced
    # -------------------------
    plt.figure(figsize=(12, 5))

    plt.subplot(1, 2, 1)
    plt.imshow(src_image, cmap='gray')
    plt.title("Original - Low Contrast")
    plt.axis('off')

    plt.subplot(1, 2, 2)
    plt.imshow(enhanced_image, cmap='gray')
    plt.title("Enhanced - Contrast Stretching")
    plt.axis('off')

    plt.tight_layout()
    plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read Image 2
src_image = cv2.imread('/content/Image2_Underexposed.jpg',
                       cv2.IMREAD_GRAYSCALE)

# Check whether image is loaded
if src_image is None:
    print("Image not found!")
else:

    # -------------------------
    # Gamma Correction
    # -------------------------
    gamma = 2.5

    normalized_image = src_image / 255.0

    gamma_corrected = np.power(normalized_image, 1 / gamma)

    enhanced_image = np.uint8(gamma_corrected * 255)

    # -------------------------
    # Display Images
    # -------------------------
    plt.figure(figsize=(12, 5))

    plt.subplot(1, 2, 1)
    plt.imshow(src_image, cmap='gray', vmin=0, vmax=255)
    plt.title("Original - Underexposed")
    plt.axis('off')

    plt.subplot(1, 2, 2)
    plt.imshow(enhanced_image, cmap='gray', vmin=0, vmax=255)
    plt.title("Enhanced - Gamma Correction")
    plt.axis('off')

    plt.tight_layout()
    plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

src_image = cv2.imread('/content/Image3_Overexposed.png',
                       cv2.IMREAD_GRAYSCALE)

if src_image is None:
    print("Image not found!")
else:

    gamma = 2.5

    normalized_image = src_image / 255.0

    enhanced_image = np.power(normalized_image, gamma)

    enhanced_image = np.uint8(enhanced_image * 255)

    plt.figure(figsize=(12, 5))

    plt.subplot(1, 2, 1)
    plt.imshow(src_image, cmap='gray', vmin=0, vmax=255)
    plt.title("Original - Overexposed")
    plt.axis('off')

    plt.subplot(1, 2, 2)
    plt.imshow(enhanced_image, cmap='gray', vmin=0, vmax=255)
    plt.title("Enhanced - Gamma Correction")
    plt.axis('off')

    plt.tight_layout()
    plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt

# Read Image 4
src_image = cv2.imread('/content/Image4_Foggy.jpg',
                       cv2.IMREAD_GRAYSCALE)

# Check whether image is loaded
if src_image is None:
    print("Image not found!")
else:

    # -------------------------
    # Histogram Equalization
    # -------------------------
    enhanced_image = cv2.equalizeHist(src_image)

    # -------------------------
    # Display Original & Enhanced
    # -------------------------
    plt.figure(figsize=(12, 5))

    plt.subplot(1, 2, 1)
    plt.imshow(src_image, cmap='gray', vmin=0, vmax=255)
    plt.title("Original - Foggy")
    plt.axis('off')

    plt.subplot(1, 2, 2)
    plt.imshow(enhanced_image, cmap='gray', vmin=0, vmax=255)
    plt.title("Enhanced - Histogram Equalization")
    plt.axis('off')

    plt.tight_layout()
    plt.show()
```
